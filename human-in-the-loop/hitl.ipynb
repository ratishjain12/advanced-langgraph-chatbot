```python
from langgraph.graph import START, StateGraph
from langchain_core.messages import BaseMessage
from langgraph.checkpoint.memory import InMemorySaver
from langgraph.graph.message import add_messages
from typing import TypedDict, Annotated
from langchain_openai import ChatOpenAI
from langgraph.types import interrupt, Command
from dotenv import load_dotenv
load_dotenv()
```

```python
class ChatState(TypedDict):
    messages: Annotated[list[BaseMessage], add_messages]
```

```python
llm = ChatOpenAI(model="gpt-5.4-nano-2026-03-17")
```

```python
def chat_node(state:ChatState)->ChatState:
    decision = interrupt(
        {
            "type": "approval",
            "message": "Do you approve the following messages to be sent to the LLM?"
        }
    )

    if decision['approved'] == 'yes':
        messages = state['messages']
        response = llm.invoke(messages)
        return {'messages': [response]}
    else:
        return {'messages': [AIMessage(content="The messages were not approved by the user.")]}
```
